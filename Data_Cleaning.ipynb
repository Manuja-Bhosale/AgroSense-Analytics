```python
#data cleaning using pandas
import pandas as pd
smart_farming_df = pd.read_csv("../raw_data/smart_Farming_Crop_Yield_2024.csv")
```

```python
#Remove duplicates from smart_farming_df
smart_farming_df.info()
```

```python
smart_farming_df = smart_farming_df.drop_duplicates()
```

```python
smart_farming_df.isnull().sum()
```

```python
smart_farming_df['irrigation_type'] = smart_farming_df['irrigation_type'].fillna('Unknown')
```

```python
smart_farming_df.isnull().sum()
```

```python
smart_farming_df['crop_disease_status'] = smart_farming_df['crop_disease_status'].fillna('None')
```

```python
smart_farming_df['sowing_date'] = pd.to_datetime(smart_farming_df['sowing_date'])
```

```python
smart_farming_df['harvest_date'] = pd.to_datetime(smart_farming_df['harvest_date'])
```

```python
smart_farming_df['timestamp'] = pd.to_datetime(smart_farming_df['timestamp'])
```

```python
smart_farming_df['crop_type'].unique()
```

```python
smart_farming_df['irrigation_type'].unique()
smart_farming_df['crop_type'].unique()
smart_farming_df['sensor_id'].unique()
smart_farming_df['region'].unique()
smart_farming_df['fertilizer_type'].unique() 
smart_farming_df['irrigation_type'].unique()
smart_farming_df['crop_disease_status'].unique() 
```

```python
#Save cleaned data of samrt_farming_df into cleaned data folder 
#Handle Missing Values(drop and filled)
#Correct Data Types(date)
#Dropped Irrelevant Columns 
#checked unique values

smart_farming_df.to_csv("../cleaned_data/cleaned_smart_farming_df_new.csv", index=False)
```
